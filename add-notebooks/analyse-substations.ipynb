```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
from shapely.wkt import loads  # Converts WKT strings to geometries

# Load the CSV file
file_path = "../../pypsa-earth/resources/DZA-48-GADM-el-netZero-2030-12H/base_network/all_lines_build_network.csv"
df = pd.read_csv(file_path)

# Check the column names
print(df.columns)

# Ensure the geometry column is correctly parsed
df["geometry"] = df["geometry"].apply(loads)  # Convert WKT strings to Shapely geometries

# Convert to a GeoDataFrame
gdf = gpd.GeoDataFrame(df, geometry="geometry", crs="EPSG:4326")  # EPSG:4326 for geographic coordinates
```

```python
gdf.explore()
```
